```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Define function and gradient
def f(x):
    return (x + 3) ** 2

def gradient(x):
    return 2 * (x + 3)
```

```python
# Initial values
x = 2
learning_rate = 0.1
iterations = 100
```

```python
# Store points for visualization
x_points = []
y_points = []
```

```python
# Gradient Descent
for i in range(iterations):
    x_points.append(x)
    y_points.append(f(x))
    x = x - learning_rate * gradient(x)
```

```python
print("Local Minimum x:", round(x, 4))
print("Minimum value y:", round(f(x), 4))
```

```python
# Plot the function
x_values = np.linspace(-6, 4, 100)
y_values = f(x_values)
```

```python
plt.plot(x_values, y_values, label='y = (x+3)^2')
plt.scatter(x_points, y_points, color='red', label='Gradient Descent')
plt.scatter(-3, 0, color='green', s=100, label='Local Minimum')

plt.title("Gradient Descent Visualization")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()
```
